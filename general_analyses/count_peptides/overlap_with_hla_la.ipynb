```python
import pandas as pd
```

```python
df = pd.read_csv("../../data/meta/peptides_by_protein.csv")
df
```

```python
hlala = pd.read_csv("HLA_la_reported_cryptic_peptides_with_metadata.csv")
hlala
```

```python
metadata = pd.read_csv("../../data/meta/20260219_EwS_ncORF_metadata_summary.tsv", sep="\t")
metadata
```

```python
hlala['coordinates'] = hlala['Location']
hlala['coordinates'] = hlala['coordinates'].apply(lambda x: x.replace('-:', ':').replace('+:', ':'))
hlala['coordinates'] = hlala['coordinates'].apply(lambda x: x.replace('|', '|'+x[:x.index(':')]+':') if '|' in x else x)
hlala['coordinates'] = hlala['coordinates'].apply(lambda x: x.split('|') if '|' in x else [x])
hlala = hlala.explode('coordinates')
hlala
```

```python
merge = metadata.merge(hlala, left_on='ORF_ranges', right_on='coordinates')
merge
```

```python
#no overlap on the coordinate level, could be because of different references used
```

```python
df_merge = df.merge(metadata, left_on='protein', right_on='orf_id')
df_merge
```

```python
'a' in 'abc'
```

```python
df_merge['detected_in_hla_la'] = False
df_merge['hla_la_peptide'] = None
for index, row in hlala.iterrows():
    for index2, row2 in df_merge.iterrows():
        if row['Sequence'] in row2['Protein']:
            df_merge.at[index2, 'detected_in_hla_la'] = True
            df_merge.at[index2, 'hla_la_peptide'] = row['Sequence']
            print(f"Match found: {row['Sequence']} in {row2['orf_id']}: {row2['Protein']}")
            
```

```python
cols = [
    'result_A673', 'result_SKNMC', 'result_MHHES',
    'result_TC106', 'result_RDES'
]

df_merge['n_downregulated'] = df_merge[cols].eq('downregulated').sum(axis=1).astype(int)
df_merge['n_downregulated'].value_counts()
```

```python
df_merge['type_of_fusion_regulation'] = df_merge['n_downregulated'].apply(lambda x: 'EF1 controlled' if x >= 3 else 'Not EF1 controlled')
df_merge['type'] = df_merge['neotranscript_orf_type'].where((df_merge['neotranscript_orf_type'] == 'Neogene ORF')|(df_merge['neotranscript_orf_type'] == 'Out-of-frame ORF')|(df_merge['neotranscript_orf_type'] == 'In-frame ORF'), df_merge['type_of_fusion_regulation'])
df_merge['type'].value_counts()
```

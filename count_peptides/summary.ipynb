```python
import pandas as pd
```

```python
df = pd.read_csv("peptides_by_protein.tsv", sep="\t")
df
```

```python
df2 = pd.read_csv("unique_peptides.csv")
df2
```

```python
dfp =pd.read_parquet('20251022_Neo_TG_vanHeesch_FF.parquet', engine='fastparquet')
dfp
```

```python
df3 = dfp[dfp['Proteotypic'] == 1.0]
df3
```

```python
dff = df3[['Protein.Group', 'Stripped.Sequence']]
dff = dff.groupby(['Protein.Group'])['Stripped.Sequence'].apply(';'.join)
dff
```

```python

d3 = pd.DataFrame({'protein':dff.index, '3_peptides_proteotypic':dff.values})
d3['protein'] = d3['protein'].apply(lambda x: x.split(';'))
d3 = d3.explode('protein')
d3
```

```python
d3['protein'] = d3['protein'].str.replace('_NEOTRANSCRIPT', '', regex=False)
```

```python
merge = df.merge(d3, how='left', on='protein')
merge
```

```python
df3 = dfp[dfp['Proteotypic'] == 0.0]
df3
```

```python
dff = df3[['Protein.Group', 'Stripped.Sequence']]
dff = dff.groupby(['Protein.Group'])['Stripped.Sequence'].apply(';'.join)
dff
```

```python

d3 = pd.DataFrame({'protein':dff.index, '3_peptides_not_proteotypic':dff.values})
d3['protein'] = d3['protein'].apply(lambda x: x.split(';'))
d3 = d3.explode('protein')
d3['protein'] = d3['protein'].str.replace('_NEOTRANSCRIPT', '', regex=False)
d3
```

```python
merge = merge.merge(d3, how='left', on='protein')
merge
```

```python
merge[['protein', '1_peptides', '1_peptides_no_match_sp', '2_peptides', '2_peptides_no_match_sp', '3_peptides_proteotypic', '3_peptides_not_proteotypic', '4_peptides', '4_peptides_no_match_sp']].to_csv('peptides_by_protein_merged.csv', index=False)
```

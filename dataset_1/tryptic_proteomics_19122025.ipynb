```python
def normalize(df):
    '''
    Returns  NSAF (Normalized Spectral Abundance Factor) instead of raws
		'''
    spectrum_count_colums = [x for x in df.columns.tolist() if 'Spectral Count' in x]
    df_nsaf = df.copy()
    for col in spectrum_count_colums:
        sample_name = col.replace(" Spectral Count","")
        df_nsaf[f'SC/L {sample_name}'] = df_nsaf[col] / df_nsaf['Protein Length']
        df_nsaf[f'{sample_name}'] = df_nsaf[f'SC/L {sample_name}'] / df_nsaf[f'SC/L {sample_name}'].sum()
    return df_nsaf
```

```python
import matplotlib.colors as mcolors
import matplotlib.cm as cm
import pandas as pd
```

```python
shap_df = pd.read_csv('shap_values.csv')
shap_df = shap_df.set_index('feature')
shap_vals = shap_df['mean_shap']  # your Series indexed like glue rows
norm = mcolors.Normalize(vmin=-shap_vals.max(), vmax=shap_vals.max())
cmap_shap = cm.get_cmap('RdBu_r')  # or 'coolwarm', 'seismic', 'bwr'

# 2. Map each SHAP value to an RGBA color
shap_colors = pd.Series(
    [mcolors.to_hex(cmap_shap(norm(v))) for v in shap_vals],
    index=shap_vals.index,
    name='SHAP value'
)
```

```python
import matplotlib.patches as mpatches

col_patches = [
    mpatches.Patch(color=color, label=label)
    for label, color in palette_cell_lines.items()
]
row_patches = [
    mpatches.Patch(color=color, label=label)
    for label, color in palette.items()
]
# 3. Combine with your existing row_colors
# row_colors can be a DataFrame with multiple columns
if isinstance(row_colors, pd.Series):
    row_colors_combined = pd.DataFrame({
        row_colors.name or 'ORF type': row_colors,
        'SHAP value': shap_colors
    })
elif isinstance(row_colors, pd.DataFrame):
    row_colors_combined = row_colors.copy()
    row_colors_combined['SHAP value'] = shap_colors
```

```python
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42   # <-- ADD THIS before creating the figure
matplotlib.rcParams['ps.fonttype'] = 42  
```

```python
g = sns.clustermap(
    glue,
    cmap='plasma',
    xticklabels=True,
    yticklabels=True,
    figsize=(10, 10),
    annot=False,
    row_colors=row_colors_combined,
    col_colors=col_colors,
    col_cluster=True,
    row_cluster=False,
    dendrogram_ratio=(0.05, 0.03),
    colors_ratio=(0.01, 0.01)
)

g.ax_heatmap.set_xticklabels(g.ax_heatmap.get_xticklabels(), fontsize=10)
g.ax_heatmap.set_yticklabels(g.ax_heatmap.get_yticklabels(), fontsize=10)
g.ax_heatmap.set_xlabel("Cell lines", fontsize=10, labelpad=10)

g.cax.set_position([0.01, 0.95, 0.015, 0.05])
g.cax.tick_params(labelsize=8)

for text in g.ax_heatmap.texts:
    text.set_fontsize(8)

pos = g.ax_row_colors.get_position()
heatmap_pos = g.ax_heatmap.get_position()
g.ax_row_colors.set_position([
    heatmap_pos.x0 - pos.width,
    pos.y0,
    pos.width,
    pos.height
])

plt.savefig(
    'tryptic_proteomics_ncORF_clustermap_27022026.pdf',
    dpi=300,
    bbox_inches='tight'
)
plt.show()

# Save legends as a separate figure
fig_leg, ax_leg = plt.subplots(figsize=(5, 5))
ax_leg.axis('off')

leg1 = ax_leg.legend(
    handles=col_patches,
    title="Cell line type",
    loc="upper left",
    bbox_to_anchor=(0, 1),
    frameon=False,
    fontsize=9,
    title_fontsize=9
)
ax_leg.add_artist(leg1)

leg2 = ax_leg.legend(
    handles=row_patches,
    title="ORF type",
    loc="upper left",
    bbox_to_anchor=(0, 0.5),
    frameon=False,
    fontsize=9,
    title_fontsize=9
)

norm = mcolors.Normalize(vmin=-shap_vals.max(), vmax=shap_vals.max())
sm = cm.ScalarMappable(cmap='RdBu_r', norm=norm)
sm.set_array([])

cbar_ax = fig_leg.add_axes([0.8, 0.12, 0.35, 0.03])  # [left, bottom, width, height]
cbar = fig_leg.colorbar(sm, cax=cbar_ax, orientation='horizontal')
cbar.set_label('SHAP value', fontsize=9)
cbar.ax.tick_params(labelsize=8)

plt.savefig('legends_27022026.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
